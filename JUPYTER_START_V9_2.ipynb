## Start server on port 8515 (do not use %run or python app.py)

```python
import sys,os,subprocess
env=os.environ.copy()
env['ANGSI_DATA_ROOT']='/data/epic20/APGCE/input_data/challenge3'
LOG=APP.parent/'server_v92.log'
with LOG.open('a') as handle:
    proc=subprocess.Popen([sys.executable,'-m','streamlit','run',str(APP),
        '--server.address=127.0.0.1','--server.port=8515','--server.headless=true',
        '--server.fileWatcherType=none'],cwd=str(APP.parent),env=env,
        stdout=handle,stderr=subprocess.STDOUT,start_new_session=True)
print('PID',proc.pid,'Log',LOG)
```

## Check server health; this checks the kernel host only

```python
import urllib.request
try:
    with urllib.request.urlopen('http://127.0.0.1:8515/_stcore/health',timeout=5) as response:
        print('Health:',response.status,response.read().decode())
except Exception as exc:
    print('Health error',repr(exc))
    print(LOG.read_text(errors='replace')[-5000:] if LOG.exists() else 'No server log')
```
