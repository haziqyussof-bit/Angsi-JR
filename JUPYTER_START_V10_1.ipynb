```python
import os,sys,subprocess,py_compile
from pathlib import Path
APP_DIR=Path('/data/epic20/APGCE/apgce_team_13/ANGSI_JR_V10_1_AUTO_VOLUMETRICS')
APP=APP_DIR/'angsi_jr_v10.py'
py_compile.compile(str(APP),doraise=True)
LOG=APP_DIR/'server_v101.log'
ENV=os.environ.copy()
ENV['ANGSI_DATA_ROOT']='/data/epic20/APGCE/input_data/challenge3'
with LOG.open('a') as f:
    proc=subprocess.Popen([sys.executable,'-m','streamlit','run',str(APP),'--server.address=127.0.0.1','--server.port=8518','--server.headless=true','--server.fileWatcherType=none'],cwd=APP_DIR,env=ENV,stdout=f,stderr=subprocess.STDOUT,start_new_session=True)
print('PID',proc.pid,'LOG',LOG)
```

```python
import urllib.request,time
for k in range(10):
    try:
        with urllib.request.urlopen('http://127.0.0.1:8518/_stcore/health',timeout=5) as r:
            print('HTTP',r.status,r.read().decode());break
    except Exception as exc:
        if k==9:print('FAILED',exc,'\n',LOG.read_text(errors='replace')[-7000:])
        else:time.sleep(1)
```
